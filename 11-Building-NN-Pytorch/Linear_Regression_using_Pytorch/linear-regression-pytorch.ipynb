Importing Dataset of California Housing 

```python

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
```

```python
housing = fetch_california_housing()
X_train_full,X_test,y_train_full,y_test = train_test_split(
    housing.data,housing.target,random_state = 42
)
```

```python
X_train,X_valid,y_train,y_valid = train_test_split(
    X_train_full,y_train_full,random_state = 42
)
```

```python
X_train = torch.FloatTensor(X_train)
X_valid = torch.FloatTensor(X_valid)
X_test = torch.FloatTensor(X_test)
mean = X_train.mean(dim = 0,keepdims =  True)
std = X_train.std(dim =0,keepdims = True)

X_train = (X_train - mean) / std
X_valid = (X_valid  - mean) / std
X_test = (X_valid  - mean) / std
```

```python
y_train = torch.FloatTensor(y_train).reshape(-1,1)
y_valid = torch.FloatTensor(y_valid).reshape(-1,1)
y_test = torch.FloatTensor(y_test).reshape(-1,1)
```

```python
X_train.shape
```

```python

n_features = X_train.shape[1]
```

## 7. Initialize Weights and Bias

In this section, we initialize the model parameters.

- Weights (W): Represent the importance of input features.
- Bias (b): Represents the intercept of the model.
- requires_grad=True: Allows PyTorch Autograd
  to calculate gradients automatically.

```python
# 2. Initialize parameters
w = torch.rand(n_features,1,requires_grad = True)
b = torch.tensor(0.0,requires_grad = True)
```

### Calculate Mean Squared Error (MSE)

We calculate the difference between predicted
values and actual target values.

The squared errors are averaged to calculate
the Mean Squared Error (MSE).

Our objective is to minimize this loss during training.

```python
# 3. Training loop
Learning_Rate = 0.1
n_epoch = 50
for epoch in range(n_epoch):
    # Forward pass
    y_pred = X_train @ w + b
    loss = ((y_pred - y_train) ** 2).mean()
    # Backward pass
    loss.backward()
    with torch.no_grad():
        # Update parameters (no grad)
        b -= Learning_Rate * b.grad
        w -= Learning_Rate * w.grad
        # Zero gradients
        b.grad.zero_()
        w.grad.zero_()
    print(f'Epoch: {epoch + 1}/{ n_epoch} loss: {loss.item()}')
```
