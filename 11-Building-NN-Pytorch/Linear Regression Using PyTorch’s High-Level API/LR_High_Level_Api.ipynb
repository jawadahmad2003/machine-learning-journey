```python

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
```

```python
housing = fetch_california_housing()
X_train_full,X_test,y_train_full,y_test = train_test_split(
    housing.data,housing.target,random_state = 42
)
```

```python
X_train,X_valid,y_train,y_valid = train_test_split(
    X_train_full,y_train_full,random_state = 42
)
```

```python
X_train = torch.FloatTensor(X_train)
X_valid = torch.FloatTensor(X_valid)
X_test = torch.FloatTensor(X_test)
mean = X_train.mean(dim = 0,keepdims =  True)
std = X_train.std(dim =0,keepdims = True)

X_train = (X_train - mean) / std
X_valid = (X_valid  - mean) / std
X_test = (X_valid  - mean) / std
```

```python
y_train = torch.FloatTensor(y_train).reshape(-1,1)
y_valid = torch.FloatTensor(y_valid).reshape(-1,1)
y_test = torch.FloatTensor(y_test).reshape(-1,1)
```

```python

n_features = X_train.shape[1]
```

```python
model = nn.Linear(in_features = n_features,out_features = 1 )
```

```python
Learning_Rate = 0.1
n_epoch = 50
```

```python
optimizer = torch.optim.SGD(model.parameters(),lr = Learning_Rate)
mse = nn.MSELoss()
```

```python
def train_bgd(model,optimizer,criterion,X_train,y_train,n_epoch):
    for epoch in range(n_epoch):
        y_pred = model(X_train)
        loss = criterion(y_pred,y_train)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        print(f'Epoch: {epoch + 1} / {n_epoch} Loss: {loss.item()}')
    
train_bgd(model,optimizer,mse,X_train,y_train,n_epoch)
```

Implementing Mini-Batch Gradient
Descent Using DataLoaders

```python
from torch.utils.data import TensorDataset,DataLoader
train_datatset = TensorDataset(X_train,y_train)
train_loader = DataLoader(train_datatset,batch_size = 32,shuffle = True,)
```

```python
model = nn.Sequential(
    nn.Linear(in_features = n_features, out_features = 50),
    nn.ReLU(),
    nn.Linear(50,40),
    nn.ReLU(),
    nn.Linear(40,1)
)
```

```python
if torch.cuda.is_available():
    device = 'cuda'
else:
    device = 'cpu'
```

```python
def train(model, optimizer, criterion, train_loader, n_epochs):
    model.train()
    for epoch in range(n_epochs):
        total_loss = 0.
        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            y_pred = model(X_batch)
            loss = criterion(y_pred, y_batch)
            total_loss += loss.item()
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()
            mean_loss = total_loss / len(train_loader)
            print(f"Epoch {epoch + 1}/{n_epochs}, Loss: {mean_loss:.4f}")
```

```python
train(model,optimizer,mse,train_loader,n_epoch)
```
