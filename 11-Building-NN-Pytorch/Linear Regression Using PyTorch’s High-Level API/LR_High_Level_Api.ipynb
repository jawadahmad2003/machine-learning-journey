```python
y_train = torch.FloatTensor(y_train).reshape(-1,1)
y_valid = torch.FloatTensor(y_valid).reshape(-1,1)
y_test = torch.FloatTensor(y_test).reshape(-1,1)
```

```python
def train_bgd(model,optimizer,criterion,X_train,y_train,n_epoch):
    for epoch in range(n_epoch):
        y_pred = model(X_train)
        loss = criterion(y_pred,y_train)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        print(f'Epoch: {epoch + 1} / {n_epoch} Loss: {loss.item()}')
    
train_bgd(model,optimizer,mse,X_train,y_train,n_epoch)
```

Implementing Mini-Batch Gradient
Descent Using DataLoaders

```python
from torch.utils.data import TensorDataset,DataLoader
train_datatset = TensorDataset(X_train,y_train)
train_loader = DataLoader(train_datatset,batch_size = 32,shuffle = True,)
```

```python
model = nn.Sequential(
    nn.Linear(in_features = n_features, out_features = 50),
    nn.ReLU(),
    nn.Linear(50,40),
    nn.ReLU(),
    nn.Linear(40,1)
)
```

```python
if torch.cuda.is_available():
    device = 'cuda'
else:
    device = 'cpu'
```

```python
def train(model, optimizer, criterion, train_loader, n_epochs):
    model.train()
    for epoch in range(n_epochs):
        total_loss = 0.
        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            y_pred = model(X_batch)
            loss = criterion(y_pred, y_batch)
            total_loss += loss.item()
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()
            mean_loss = total_loss / len(train_loader)
            print(f"Epoch {epoch + 1}/{n_epochs}, Loss: {mean_loss:.4f}")
```

```python
train(model,optimizer,mse,train_loader,n_epoch)
```

Model Evaluation


```python
def evaluate(model,data_loader,metric_fn,aggregate_fn = torch.mean):
    model.eval()
    metric_values = []
    with torch.no_grad():
        for X_batch, y_batch in data_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            y_pred = model(X_batch)
            metric_value = metric_fn(y_pred, y_batch)
            metric_values.append(metric_value)
    return aggregate_fn(torch.tensor(metric_values))
```

```python
valid_dataset = TensorDataset(X_valid,y_valid)
valid_loader = DataLoader(valid_dataset,batch_size = 32)
valid_mse = evaluate(model,valid_loader,mse)

valid_mse
```

```python
def rmse(y_pred, y_true):
    return torch.sqrt(mse(y_pred, y_true))

evaluate(model, valid_loader, rmse)
```

Building Nonsequential Models Using Custom Modules



```python
#Pytorch Base class for Neural Network Creation nn.Module
class WideandDeepModel(nn.Module):
    #Constructor init model initialize karta hai 
    #self current model object represent karta hai
    #n_features input features ki number hai
    #super() parent class ka constructor call karta hai
    def __init__(self,n_features):
        super().__init__()
        self.deep_stack = nn.Sequential(
            nn.Linear(n_features, 50),
            nn.ReLU(),
            nn.Linear(50, 40),
             
        )
        self.output = nn.Linear(40 + n_features, 1)

    def forward(self, X):
        deep_out = self.deep_stack(X)
        wide_and_deep = torch.cat([X, deep_out], dim=1)
        return self.output(wide_and_deep)
```

```python
torch.manual_seed(42)
model = WideandDeepModel(n_features)
```

```python
Learning_Rate = 0.001
n_epoch = 50
train(model, optimizer, mse, train_loader, n_epoch)
```
