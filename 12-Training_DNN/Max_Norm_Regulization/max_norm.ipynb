```python
import torch
import torch.nn as nn

class MaxNormModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.hidden = nn.Linear(784, 100)
        self.output = nn.Linear(100, 10)

    def forward(self, x):
        x = torch.relu(self.hidden(x))
        return self.output(x)

    def apply_max_norm(self, max_norm=3.0):

        with torch.no_grad():

            self.hidden.weight.copy_(
                torch.renorm(
                    self.hidden.weight,
                    p=2,
                    dim=0,
                    maxnorm=max_norm
                )
            )
```
