```python
import torch
import torch.nn as nn
import torch.optim as optim

# Neural Network
model = nn.Sequential(
    nn.Linear(784, 100),
    nn.ReLU(),
    nn.Linear(100, 10)
)

# Loss function
criterion = nn.CrossEntropyLoss()

# Optimizer
optimizer = optim.Adam(model.parameters(), lr=0.001)

# L1 Regularization
l1_lambda = 0.01

# Example training batch
X = torch.randn(32, 784)
y = torch.randint(0, 10, (32,))

# Forward pass
y_pred = model(X)

# Original loss
original_loss = criterion(y_pred, y)

# Calculate L1 penalty
l1_penalty = sum(
    torch.sum(torch.abs(param))
    for param in model.parameters()
    if param.requires_grad
)

# Total loss
loss = original_loss + l1_lambda * l1_penalty

# Backpropagation
optimizer.zero_grad()
loss.backward()
optimizer.step()
```
